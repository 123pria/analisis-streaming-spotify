Visualisasi Data Dengan Matplotlib

```python
df["Sex"].value_counts().plot(kind="pie", autopct="%1.1f%%", figsize=(5, 5))

plt.title("Distribution of Artists by Sex")
plt.xlabel("Sex")
plt.ylabel("Number of Artists")
plt.show()
```

```python
country_counts = df["Country of Origin"].value_counts().head(10)

country_counts.plot(kind="bar", color=["green","olive","purple"],figsize=(6, 5))

plt.title("Top 10 Countries by Number of Artists")
plt.xlabel("Country")
plt.ylabel("Number of Artists")

plt.xticks(rotation=45)

plt.show()
```

```python
genre_counts = df["Primary Genre"].value_counts().head(10)

genre_counts.plot(kind="bar",figsize=(6, 4), color=["blue","orange","green","red","purple","brown","pink","gray","olive","cyan"])

plt.title("Top 10 Music Genres")
plt.xlabel("Genre")
plt.ylabel("Number of Artists")

plt.xticks(rotation=45)

plt.show()
```

```python
plt.hist(df["Total Streams (in millions)"])

plt.title("Distribution of Total Streams")
plt.xlabel("Total Streams (Millions)")
plt.ylabel("Number of Artists")

plt.show()
```

```python
sns.boxplot(
    x=df["Total Streams (in millions)"]
)

plt.title("Distribution of Total Streams")

plt.show()
```

```python
df.groupby("Primary Genre")[
    "Total Streams (in millions)"
].mean()
```

```python
genre_streams = (
    df.groupby("Primary Genre")[
        "Total Streams (in millions)"
    ]
    .mean()
    .sort_values(ascending=False)
)
```

```python
genre_streams.head()
```

```python
df.groupby("Sex")[
    "Total Streams (in millions)"
].mean()
```

```python
df.groupby("Country of Origin")[
    "Total Streams (in millions)"
].mean().sort_values(ascending=False).head(10)
```

```python
stream_columns = [
    "Total Streams (in millions)",
    "Lead Streams (in millions)",
    "Feature Streams (in millions)",
    "Solo Streams (in millions)",
    "Collaborative Streams (in millions)"
]

df[stream_columns].mean()
```

```python
df[stream_columns].mean().plot(kind="bar")

plt.title("Average Streams by Stream Type")
plt.xlabel("Stream Type")
plt.ylabel("Average Streams (Millions)")

plt.xticks(rotation=45)

plt.show()
```

```python
numeric_columns = [
    "Debut Year",
    "Total Streams (in millions)",
    "Lead Streams (in millions)",
    "Feature Streams (in millions)",
    "Solo Streams (in millions)",
    "% of Solo Streams",
    "Collaborative Streams (in millions)",
    "% of Collaborative Streams"
]
```

```python
correlation = df[numeric_columns].corr()
```

```python
correlation
```

```python
plt.figure(figsize=(10, 5))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Between Spotify Variables")

plt.show()
```
